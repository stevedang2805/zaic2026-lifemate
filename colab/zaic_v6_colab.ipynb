# ZAIC 2026 LifeMate — v6: quay lại công thức v4 + dữ liệu synthetic

**v5 làm điểm tụt: 44.46 → 40.49.** v5 đổi hai thứ cùng lúc so với v4 và cả hai đều đáng ngờ:
`--detach-prefix` (LoRA không nhận gradient từ phần prompt, nên học đọc schema kém đi) và các bản transcript ASR
(khi ASR nghe sai một con số mà đáp án vẫn là con số đúng, model học cách **đoán bừa thay vì hỏi lại** —
trên public, số lần gọi `sys.ask_missing_info` tụt từ 92 xuống 62, còn lỗi sai đối số thì tăng).

v6 quay lại đúng công thức của v4 và **chỉ thêm một thay đổi: dữ liệu synthetic**.

**Bắt buộc:** đặt các file spec vào `MyDrive/ZAIC/synth_specs/` trước khi chạy, nếu không notebook sẽ dừng.
GPU **A100**, khoảng 2.5–3 giờ.

```python
# ==== CẤU HÌNH ====
BASE_MODEL = "Qwen/Qwen3-4B-Instruct-2507"
BATCH      = 32
LLM_EPOCHS = 1
TRAIN_ALL  = False    # False: train trên train_fit, giữ val để đo (so sánh sạch với v4)
                      # True : train trên cả 1545 mẫu — chỉ bật cho lần train CUỐI, khi v6 đã chứng minh tốt hơn
USE_ASR    = False    # giữ False. True: thêm bản transcript ASR đã lọc (chỉ để thử A/B khi còn thời gian)
DATA_URL   = "https://dl-challenge.zalo.ai/2026/lifemate_argfzdgert/train+public_test.zip"
DRIVE_DIR  = "/content/drive/MyDrive/ZAIC"
ASR_PROD   = DRIVE_DIR + "/asr_ft_v4"
WORK, DATA_RAW, RES = "/content/work", "/content/data_raw", "/content/results_v6"
LLM_OUT    = "/content/ft_v6"
```

```python
import subprocess, torch, os
print(subprocess.run(["nvidia-smi"], capture_output=True, text=True).stdout)
assert torch.cuda.is_available(), "Chưa bật GPU"
DTYPE = "bfloat16" if torch.cuda.get_device_capability()[0] >= 8 else "float16"
GPU_NAME = torch.cuda.get_device_name(0); print(GPU_NAME, DTYPE)
os.makedirs(RES, exist_ok=True)
```

```python
from google.colab import drive
drive.mount("/content/drive")
import os, zipfile, shutil, glob, subprocess, json, time
shutil.rmtree(WORK, ignore_errors=True); zipfile.ZipFile(os.path.join(DRIVE_DIR, "lifemate_baseline.zip")).extractall(WORK)
SPEC_DIR = os.path.join(DRIVE_DIR, "synth_specs")
specs = sorted(glob.glob(SPEC_DIR + "/*.jsonl"))
assert specs, "Chưa có file spec nào trong " + SPEC_DIR + " — v6 chỉ có nghĩa khi đã có dữ liệu synthetic"
print("synth spec files:", len(specs), [os.path.basename(p) for p in specs])
data_zip = os.path.join(DRIVE_DIR, "train+public_test.zip")
if not os.path.exists(data_zip):
    data_zip = "/content/train+public_test.zip"
    if not os.path.exists(data_zip):
        assert subprocess.run(["wget", "-q", "-O", data_zip, DATA_URL]).returncode == 0, "Tải dữ liệu lỗi"
if not glob.glob(DATA_RAW + "/**/train/train.jsonl", recursive=True):
    zipfile.ZipFile(data_zip).extractall(DATA_RAW)
DATA = os.path.dirname(os.path.dirname(glob.glob(DATA_RAW + "/**/train/train.jsonl", recursive=True)[0]))
IDS  = "" if TRAIN_ALL else "--ids splits/train_fit.txt"
ASRF = "--asr asr_train_oof.json" if USE_ASR else ""
if USE_ASR:
    shutil.copy(os.path.join(DRIVE_DIR, "asr_train_oof.json"), WORK + "/asr_train_oof.json")
print("DATA:", DATA)
```

## 1. Dựng và kiểm tra dữ liệu synthetic bằng simulator của BTC

```python
%cd {WORK}
!python -m lifemate.synth_build --root {DATA} --tools-from public_test --specs {SPEC_DIR}/*.jsonl 2>&1 | tee {RES}/synth_build.log | tail -25
```

```python
import collections as C
rows = [json.loads(l) for l in open(f"{DATA}/synth/synth.jsonl", encoding="utf-8")]
tag = lambda r: "oos" if not r["label"]["tool_calls"] else ("ask" if any(c["name"].startswith("sys.") for c in r["label"]["tool_calls"]) else "act")
apps = C.Counter(a for r in rows for a in r["context"]["installed_apps"])
n_synth = len(rows)
print("mẫu synthetic dựng được:", n_synth, "|", dict(C.Counter(tag(r) for r in rows)))
print("app:", apps.most_common(14))
assert n_synth >= 50, "quá ít mẫu qua được kiểm tra — xem synth_build.log rồi sửa spec"
open(f"{RES}/synth_stats.txt", "w").write(f"{n_synth} mẫu\n{dict(C.Counter(tag(r) for r in rows))}\n{apps.most_common()}\n")
```

## 2. Dữ liệu SFT (công thức v4: bản gốc + bản đổi tên app)

```python
%cd {WORK}
!python -m lifemate.sft_data --root {DATA} {IDS} --aug-copies 1 {ASRF} --out sft_real.jsonl
!python -m lifemate.sft_data --root {DATA} --split synth --aug-copies 1 --out sft_synth.jsonl
with open("sft_all.jsonl", "w", encoding="utf-8") as out:
    for p in ["sft_real.jsonl", "sft_synth.jsonl"]:
        out.writelines(open(p, encoding="utf-8"))
n = sum(1 for _ in open("sft_all.jsonl")); print("tổng ví dụ SFT:", n)
```

```python
%cd {WORK}
!python -m lifemate.train_lora --data sft_all.jsonl --model {BASE_MODEL} --out {LLM_OUT} --epochs {LLM_EPOCHS} 2>&1 | grep -vi warn
assert os.path.exists(f"{LLM_OUT}/merged/config.json"), "train lỗi: xem log phía trên"
shutil.copy(f"{LLM_OUT}/train_log.txt", f"{RES}/train_log.txt")
shutil.copytree(f"{LLM_OUT}/adapter", os.path.join(DRIVE_DIR, "ft_v6_adapter"), dirs_exist_ok=True)
```

## 4. Đánh giá trên val với ASR thật (v4: 35.06 / 54.11 — mốc cần vượt)

```python
%cd {WORK}
!python -m lifemate.asr --root {DATA} --split train --ids splits/val_all.txt --model {ASR_PROD} --dtype {DTYPE} --out asr_val.json 2>&1 | grep -E "^model=|^WER"
sc = {}
for S in ["val_unseen", "val_seen"]:
    !python -m lifemate.run --root {DATA} --split train --ids splits/{S}.txt --backend vllm --model {LLM_OUT}/merged --dtype {DTYPE} --batch {BATCH} --transcripts asr_val.json --out runs/v6_{S} 2>&1 | tail -1
    !python scripts/score.py --root {DATA} --split train --pred runs/v6_{S}/submission.jsonl --by-tag > {RES}/score_v6_{S}.txt
    !python -m lifemate.report --root {DATA} --split train --pred runs/v6_{S}/submission.jsonl --show 6 > {RES}/report_v6_{S}.txt
    sc[S] = float(open(f"{RES}/score_v6_{S}.txt").readline().split()[-1])
    !head -9 {RES}/report_v6_{S}.txt
print("v6:", sc, "| v4: {'val_unseen': 35.06, 'val_seen': 54.11} | v5: {'val_unseen': 37.67, 'val_seen': 52.96}")
if TRAIN_ALL:
    print("CHÚ Ý: TRAIN_ALL=True nên val đã nằm trong tập train — điểm này cao giả tạo, chỉ nhìn điểm public")
json.dump(sc, open(f"{RES}/e2e_v6.json", "w"))
```

## 5. Public → `submission_public_v6.jsonl` (mốc cần vượt: **44.46**)

```python
%cd {WORK}
!python -m lifemate.asr --root {DATA} --split public_test --model {ASR_PROD} --dtype {DTYPE} --out asr_public.json 2>&1 | tail -1
!python -m lifemate.run --root {DATA} --split public_test --backend vllm --model {LLM_OUT}/merged --dtype {DTYPE} --batch {BATCH} --transcripts asr_public.json --out runs/public_v6 2>&1 | tail -1
shutil.copy("runs/public_v6/submission.jsonl", f"{RES}/submission_public_v6.jsonl")
shutil.copy("runs/public_v6/submission.jsonl", os.path.join(DRIVE_DIR, "submission_public_v6.jsonl"))
shutil.copy("runs/public_v6/traces.jsonl", f"{RES}/traces_public_v6.jsonl")
shutil.copy("asr_public.json", f"{RES}/asr_public.json")
sub = [json.loads(l) for l in open("runs/public_v6/submission.jsonl")]
ask = sum(1 for x in sub if any(c["name"].startswith("sys.") for c in x["tool_calls"]))
empty = sum(1 for x in sub if not x["tool_calls"])
print(f"dòng {len(sub)} | ask_missing_info {ask} (v4: 92, v5: 62) | rỗng {empty} (v4: 64, v5: 66)")
```
